## Análise de dataset

A empresa Umbrella Corporation contratou desenvolvedores Python (🐍) para analisar um dataset para verificar o alto número de cancelamentos de contratos dessa empresa, e para entregar uma solução para este programa.

Para isso, a Umbrella Corp. forneceu um *dataset* com os detalhes dos cancelamentos. Os Pythonianos (🐍) devem analizar o arquivo utilizado o **Pandas** (🐼).
